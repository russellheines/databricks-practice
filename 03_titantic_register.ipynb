# Titanic Survival Prediction
Example of registering models with MLflow.

```python
import mlflow
```

```python
# Fetch the latest successful runs from your experiment
# (Or hardcode the specific Run IDs you want to register)
experiment = mlflow.get_experiment_by_name("/Shared/titanic_experiment")
runs = mlflow.search_runs(experiment_ids=[experiment.experiment_id])

# Extract the run IDs based on the unique names you gave them
sklearn_run_id = runs[runs["tags.mlflow.runName"] == "titanic_sklearn_run"].iloc[0]["run_id"]
sparkml_run_id = runs[runs["tags.mlflow.runName"] == "titanic_sparkml_run"].iloc[0]["run_id"]
```

```python
# 1. Register the scikit-learn model
mlflow.register_model(
    model_uri=f"runs:/{sklearn_run_id}/model",
    name="titanic_sklearn_model"
)
```

```python

# 2. Register the SparkML model
mlflow.register_model(
    model_uri=f"runs:/{sparkml_run_id}/model",
    name="titanic_sparkml_model"
)
```
