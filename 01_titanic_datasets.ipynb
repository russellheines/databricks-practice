# Titanic Survival Prediction
Example of creating Delta tables with data from csv files.

```python
# 1. Read the raw CSV data into Spark DataFrames
train_df = (spark.read
            .format("csv")
            .option("header", "true")       # Uses the first row as column headers
            .option("inferSchema", "true")  # Automatically sets columns like Age to numbers
            .load("/Workspace/Users/russellheines@gmail.com/databricks-practice/train.csv"))

test_df = (spark.read
           .format("csv")
           .option("header", "true")
           .option("inferSchema", "true")
           .load("/Workspace/Users/russellheines@gmail.com/databricks-practice/test.csv"))

# 2. Write the DataFrames out as permanent Delta Tables
# Because this is Community Edition, these save directly to hive_metastore.default
train_df.write.format("delta").mode("overwrite").saveAsTable("titanic_train")
test_df.write.format("delta").mode("overwrite").saveAsTable("titanic_test")
```
